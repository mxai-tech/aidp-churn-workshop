Plantilla pública para configurar una conexión segura. No contiene secretos ni datos operativos. El notebook completo se entrega por el canal indicado por el instructor.

```python
import os

MYSQL_CONFIG = {
    'host': os.environ.get('MYSQL_HOST'),
    'port': os.environ.get('MYSQL_PORT', '3306'),
    'database': os.environ.get('MYSQL_DATABASE'),
    'user': os.environ.get('MYSQL_USER'),
    'password': os.environ.get('MYSQL_PASSWORD'),
}

missing = [key for key, value in MYSQL_CONFIG.items() if not value]
if missing:
    raise ValueError(f'Faltan variables de entorno: {missing}')

jdbc_url = (
    f"jdbc:mysql://{MYSQL_CONFIG['host']}:{MYSQL_CONFIG['port']}/"
    f"{MYSQL_CONFIG['database']}?useSSL=false"
)
```
